CCS-N（Cumulative Context Score，累积上下文得分）的计算方式如下：

1. **输入**：对每个SNV，分别构建参考序列和突变序列。
2. **逐token打分**：将两条序列输入同一DNA基础模型，在嵌入空间或概率空间中，计算每个对应token之间的零样本距离/差异得分。  
   - 嵌入空间：欧氏距离、曼哈顿距离、余弦相似度等  
   - 概率空间：Jensen-Shannon散度、Hellinger距离、交叉熵等
3. **确定中心**：以变异所在token为中心，记为 \(s_0\)。
4. **对称累加邻近token**：  
   \[
   CCS\text{-}N = s_0 + s_1^+ + s_1^- + s_2^+ + s_2^- + \dots + s_N^+ + s_N^-
   \]  
   其中 \(s_i^+\) 和 \(s_i^-\) 分别表示中心右侧和左侧第 \(i\) 个token的得分。
5. **含义**：将中心及左右各N个邻近token的“参考 vs 突变”差异得分直接求和，用于放大变异对局部序列上下文的累积扰动。
6. **特点**：不是Attention，不是相关性，不接下游任务头；最优N通常为3或4。
