````python
class MultiTrackPredictor:
    """与 notebook 中相同的预测器类"""
    def __init__(self, fasta_path: str, base_model_path: str, sft_ckpt_path: str,
                 tokenizer_path: str, use_flash_attn: bool, index_stat_path: str,
                 **kwargs):
        self.fasta = pyfaidx.Fasta(fasta_path)
        self.tokenizer = AutoTokenizer.from_pretrained(tokenizer_path)
        self.model = load_finetuned_model(
            model_class=GenOmics,
            model_path=base_model_path,
            ckpt_path=sft_ckpt_path,
            use_flash_attn=use_flash_attn,
            device="cuda:0",
            model_init_kwargs={"index_stat": json.load(open(index_stat_path, "r")),
                               **kwargs}
        )
        print("✅ Model loaded successfully.")
        print(self.model)
        self.model.eval()

    def predict(self, chrom: str, start: int, end: int, biosample_names: list = None) -> dict:
        predict_sequence = load_fasta_sequence(self.fasta, chrom, start, end)
        inputs = self.tokenizer(
            predict_sequence,
            return_tensors="pt",
            padding=False,
            truncation=True,
            max_length=32768,
            add_special_tokens=False
        ).to("cuda")
        with torch.no_grad():
            start_time = time.time()
            result = self.model.predict(inputs['input_ids'],
                                        biosample_names=biosample_names)
            time_taken = time.time() - start_time
            torch.cuda.empty_cache()
        print(f"Inference time: {time_taken:.2f} s.")
        return {
            'sequence': predict_sequence,
            'position': (chrom, start, end),
            'values': result
        }

    def predict2(self, chrom: str, start: int, end: int, seq: str, biosample_names: list = None) -> dict:
        sequence = pyfaidx.Fasta(seq)
        predict_sequence = str(sequence[0][:])
        inputs = self.tokenizer(
            predict_sequence,
            return_tensors="pt",
            padding=False,
            truncation=True,
            max_length=32768,
            add_special_tokens=False
        ).to("cuda")
        with torch.no_grad():
            start_time = time.time()
            result = self.model.predict(inputs['input_ids'],
                                        biosample_names=biosample_names)
            time_taken = time.time() - start_time
            torch.cuda.empty_cache()
        print(f"Inference time: {time_taken:.2f} s.")
        return {
            'sequence': predict_sequence,
            'position': (chrom, start, end),
            'values': result
        }
    

def extract_example_name(mut_fasta_path: str) -> str:
    """从突变 FASTA 文件路径中提取示例名称。
    假设文件名格式为 'example_XXX.alt.fa' 或类似，返回 'example_XXX'。
    """
    basename = Path(mut_fasta_path).stem  # 去掉扩展名，例如 'example_106.alt'
    match = re.search(r'(example_\w+)', basename)
    if match:
        return match.group(1)
    else:
        return basename
````

````python
# ==================== 全自动：NIP基因序列 → 比对P7 → BW可视化 ====================
"""
流程:
  1)  从 NIP 参考基因组提取 LOC_Os09g35980 序列 (Chr9:20731589-20734728)
  2)  minimap2 将该序列比对到 P7-new.fasta
  3)  从 PAF 取最长比对区域
  4)  从 P7_EVM.all.gff3 提取该区域内的基因坐标
  5)  上下游各 padding 2kb
  6)  映射染色体名 → BigWig 格式
  7)  加载 CSQ_P7_1.bw 信号并绘图（含外显子标注）
"""
import subprocess, re, tempfile
from src.dataset import load_bigwig_signal, load_fasta_sequence
from src.viewer import gaussian_smooth

MINIMAP2 = "/mnt/rice/default/Workspace/xuxiaolong/mamba/envs/liftover/bin/minimap2"

# ── 1. 提取 NIP 基因序列 ──
NIP_FASTA = fasta_path  # osa1_r7.asm.ch.fa
REF_FASTA = "/mnt/rice/default/Workspace/yangdong/18k/ref/P7-new.fasta"
GFF_PATH  = "/mnt/rice/default/Workspace/yangdong/18k/ref/P7_EVM.all.gff3"
BW_PATH   = "/mnt/rice/default/Workspace/yangdong/gene_expression_prediction/data2/CSQ_4_12/processed/renorm_bigwig_output/CSQ_P7_1.bw"
PADDING   = 2000

GENE_NAME = "LOC_Os09g35980"
NIP_CHROM = "Chr9"
NIP_START = 20731589  # 0-based for load_fasta_sequence
NIP_END   = 20734728

print(f">>> Extracting {GENE_NAME} sequence from NIP ({NIP_CHROM}:{NIP_START}-{NIP_END}) ...")
import pyfaidx
nip_fasta = pyfaidx.Fasta(NIP_FASTA)
gene_seq = str(nip_fasta[NIP_CHROM][NIP_START:NIP_END])
nip_fasta.close()
print(f"    Sequence length: {len(gene_seq)} bp")

# 写入临时文件
tmp_fa = tempfile.NamedTemporaryFile(mode='w', suffix='.fa', delete=False, prefix='nip_gene_')
tmp_fa.write(f">{GENE_NAME} NIP chr9:{NIP_START}-{NIP_END}\n{gene_seq}\n")
tmp_fa.close()
QUERY_FASTA = tmp_fa.name

# ── 2. minimap2 比对 ──
print(f">>> Aligning NIP gene sequence to P7-new.fasta via minimap2 ...")
result = subprocess.run(
    [MINIMAP2, "-x", "asm5", REF_FASTA, QUERY_FASTA],
    capture_output=True, text=True)

paf_lines = [l for l in result.stdout.strip().split("\n") if l]
if not paf_lines:
    raise RuntimeError("minimap2 未产生任何比对结果")

# ── 3. 解析 PAF，选最长比对 ──
best = None
best_len = 0
for line in paf_lines:
    cols = line.split("\t")
    rname = cols[5]       # 参考序列名: "Chr01_ZhongZao35_44597794"
    rstart = int(cols[7]) # 0-based
    rend   = int(cols[8])
    alen   = int(cols[10])# 比对长度
    if alen > best_len:
        best_len = alen
        best = (rname, rstart, rend, alen)

if best is None:
    raise RuntimeError("无法找到有效比对")
rname, rstart_g, rend_g, alen = best
print(f"Best hit: {rname}:{rstart_g}-{rend_g}  (alen={alen} bp)")

# ── 4. 从 GFF 提取该区域内的基因 ──
print(f">>> Reading genes from GFF overlapping {rname}:{rstart_g}-{rend_g} ...")
genes_in_region = []  # (gene_id, chrom, start, end, strand, name)
with open(GFF_PATH) as f:
    for line in f:
        if line.startswith("#") or not line.strip():
            continue
        cols = line.rstrip().split("\t")
        if len(cols) < 9:
            continue
        chrom_gff, _, ftype, gs, ge, _, strand, _, attrs = cols[:9]
        if ftype != "gene":
            continue
        if chrom_gff != rname:
            continue
        gs, ge = int(gs), int(ge)
        # 检查重叠
        if ge < rstart_g or gs > rend_g:
            continue
        # 提取 gene name
        name = ""
        for tag in attrs.split(";"):
            tag = tag.strip()
            if tag.startswith("Name="):
                name = tag.split("=", 1)[-1].replace("%20", " ")
                break
        gid = ""
        for tag in attrs.split(";"):
            tag = tag.strip()
            if tag.startswith("ID="):
                gid = tag.split("=", 1)[-1]
                break
        genes_in_region.append((gid, chrom_gff, gs, ge, strand, name))

if not genes_in_region:
    print("⚠️  比对区域中未找到基因注释，将使用比对区间本身")
    gene_start = rstart_g
    gene_end   = rend_g
    gene_name  = "aligned_region"
    gene_strand = "."
else:
    # 取覆盖比对齐区域的基因（或取最近的）
    genes_in_region.sort(key=lambda x: x[2])  # start
    gid, chrom_gff, gene_start, gene_end, gene_strand, gene_name = genes_in_region[0]
    print(f"Found gene: {gene_name} ({gid}) at {gene_start}-{gene_end} strand={gene_strand}")
    if len(genes_in_region) > 1:
        # 如果有多个基因重叠，合并区间
        gene_start = min(g[2] for g in genes_in_region)
        gene_end   = max(g[3] for g in genes_in_region)
        gene_name  = "+".join(g[5] or g[0] for g in genes_in_region)

# ── 5. 添加 padding ──
window_start_plot = max(0, gene_start - PADDING)
window_end_plot   = gene_end + PADDING
print(f"Plot window: {rname}:{window_start_plot}-{window_end_plot}  (gene±{PADDING}bp)")

# ── 6. 染色体名映射到 BigWig ──
chr_padded = re.sub(r"^Chr(\d+)_.*", r"Chr\1", rname)
import pyBigWig
bw_tmp = pyBigWig.open(BW_PATH)
bw_chroms = list(bw_tmp.chroms().keys())
bw_tmp.close()
if rname in bw_chroms:
    bw_chrom = rname
else:
    matches = [c for c in bw_chroms if c.startswith(chr_padded)]
    bw_chrom = matches[0] if matches else rname
    print(f"BW chromosome mapped: {rname} → {bw_chrom}")

# ── 7. 加载 BigWig 信号 ──
bw_signal = load_bigwig_signal(BW_PATH, bw_chrom, window_start_plot, window_end_plot)
display_len = window_end_plot - window_start_plot
assert len(bw_signal) == display_len, f"BW length mismatch: {len(bw_signal)} vs {display_len}"
print(f"BW signal: len={len(bw_signal)}, "
      f"range=[{bw_signal.min():.4f}, {bw_signal.max():.4f}], "
      f"nonzero={(bw_signal>0).mean()*100:.1f}%")

# ── 8. 从 GFF 提取目标基因的所有外显子（用于绘图标注） ──
gene_exons = []  # [(start, end)]
gene_id_for_exon = genes_in_region[0][0] if genes_in_region else ""
with open(GFF_PATH) as f:
    for line in f:
        if line.startswith("#") or not line.strip():
            continue
        cols = line.rstrip().split("\t")
        if len(cols) < 9:
            continue
        chrom_gff, _, ftype, es, ee, _, _, _, attrs = cols[:9]
        if ftype != "exon":
            continue
        parent = ""
        for tag in attrs.split(";"):
            tag = tag.strip()
            if tag.startswith("Parent="):
                parent = tag.split("=", 1)[-1]
                break
        if parent == gene_id_for_exon or (gene_name and f"Name={gene_name}" in attrs):
            gene_exons.append((int(es), int(ee)))
if gene_exons:
    print(f"Exons for gene: {gene_exons}")
else:
    print("No exons found for this gene, will mark the whole gene region instead")

# ── 9. 平滑 ──
smooth_sigma = getattr(args, 'smoothing_sigma', 10)
bw_smooth = gaussian_smooth(bw_signal, smooth_sigma) if smooth_sigma else bw_signal

# ── 10. 绘图 ──
x = np.arange(window_start_plot, window_end_plot)
fig, ax = plt.subplots(figsize=(14, 3.5))

ax.plot(x, bw_smooth, color='#ef822f', linewidth=1.2, alpha=0.85, label=f'BigWig (CSQ_P7_1)')
ax.fill_between(x, 0, bw_smooth, alpha=0.12, color='#ef822f')

# 外显子标注
ymax = bw_smooth.max() * 1.4 if len(bw_smooth) > 0 else 1
ax.set_ylim(0, ymax)

exon_regions = gene_exons if gene_exons else [(gene_start, gene_end)]
for es, ee in exon_regions:
    if ee > window_start_plot and es < window_end_plot:
        ds = max(es, window_start_plot)
        de = min(ee, window_end_plot)
        ax.axvspan(ds, de, color='#75bd42', alpha=0.12, linewidth=0)
        ax.plot([ds, de], [ymax*0.95, ymax*0.95],
                color='#75bd42', linewidth=3, solid_capstyle='butt')

# 基因区间标注（粗线）
ax.plot([gene_start, gene_end], [ymax*0.90, ymax*0.90],
        color='#2d6a4f', linewidth=4, solid_capstyle='butt', alpha=0.6)

# 基因名
ax.text((gene_start+gene_end)/2, ymax*0.88, gene_name,
        ha='center', va='bottom', fontsize=10, fontweight='bold',
        bbox=dict(boxstyle='round,pad=0.3', facecolor='white', alpha=0.85, edgecolor='#2d6a4f'))

ax.set_xlabel(f'Position ({rname})', fontsize=11)
ax.set_ylabel('Signal', fontsize=11)
ax.legend(fontsize=10, framealpha=0.9, edgecolor='gray')
ax.grid(True, alpha=0.2, linestyle='--', linewidth=0.5)
ax.xaxis.set_major_formatter(FuncFormatter(lambda xp, pos: f'{int(xp):,}'))

title = f'Gene {gene_name} — CSQ_P7_1 ({rname}:{window_start_plot:,}-{window_end_plot:,})'
ax.set_title(title, fontsize=12)
plt.tight_layout()
plt.show()

# ── 汇总输出 ──
print(f"\n{'='*60}")
print(f"SUMMARY")
print(f"{'='*60}")
print(f"Query seq  : {GENE_NAME} from NIP ({NIP_CHROM}:{NIP_START}-{NIP_END})")
print(f"Best match : {rname}:{rstart_g}-{rend_g} (alen={alen}bp)")
print(f"Gene       : {gene_name} ({gene_id_for_exon})")
print(f"Gene coord : {gene_start}-{gene_end} ({gene_strand})")
print(f"Plot window: {rname}:{window_start_plot}-{window_end_plot} ({display_len}bp)")
print(f"BW channel : {bw_chrom}")
print(f"{'='*60}")

# 清理临时文件
import os
os.unlink(QUERY_FASTA)
````
