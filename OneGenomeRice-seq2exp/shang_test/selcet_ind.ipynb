```python
# df_sorted = df.sort_values(by='SR_B3', ascending=False).reset_index(drop=True)
df = df.sort_values(by='SR_B3', ascending=False).reset_index(drop=True)
df.head()
```

```python
import numpy as np
import pandas as pd

# 1. 定义按 20% 间隔（5个等分点）抽样的函数
def sample_5_quantiles(group_df):
    n = len(group_df)
    if n == 0:
        return group_df
    # 计算 0%, 20%, 40%, 60%, 80% 对应的行索引位置
    indices = [int(np.round(p * (n - 1))) for p in [0.0, 0.2, 0.4, 0.6, 0.8]]
    return group_df.iloc[indices]

# 2. 筛选出两个主要亚种（注意处理字符串中可能的非打破空格 \xa0）
subspecies = ['O. sativa indica', 'O. sativa japonica']

# 清理列名两端空格/特殊空格以防匹配不上
df['Species_Clean'] = df['Species information'].str.strip()

# 3. 分组抽取并合并
df_subset = (
    df[df['Species_Clean'].isin(subspecies)]
    .groupby('Species_Clean', group_keys=False)
    .apply(sample_5_quantiles)
    .reset_index(drop=True)
)

# 查看提取出的 10 个样本及其在 df 中的分布
print("=== 抽样得到的 df_subset (共 10 个样本) ===")
print(df_subset[['Species information', 'SR_B3']])
```
