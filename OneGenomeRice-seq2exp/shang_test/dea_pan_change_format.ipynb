```python
import os
import re

# 1. 样本列表定义
japonica = ["NH001", "NH002", "NH003", "NH005", "NH013"]
indica = ["NH006", "NH007", "NH008", "NH009", "NH014"]
all_samples = japonica + indica

# 2. 目录路径定义
dirs = {
    "genome": "/mnt/rice/data/OsGenos/251cellresearch/chromosome_level/genome",
    "annotation": "/mnt/rice/default/Workspace/xuxiaolong/RNAprediction/RNA_bam/liftoff_IRGSP_full/annotations",
    "leaf_ck": "/mnt/rice/default/Workspace/xuxiaolong/RNAprediction/RNA_bam/analysis/six_context_global_tmm_bigwig/bigwig/leaf_ck",
    "leaf_salt": "/mnt/rice/default/Workspace/xuxiaolong/RNAprediction/RNA_bam/analysis/six_context_global_tmm_bigwig/bigwig/leaf_salt",
    "root_ck": "/mnt/rice/default/Workspace/xuxiaolong/RNAprediction/RNA_bam/analysis/six_context_global_tmm_bigwig/bigwig/root_ck",
    "root_salt": "/mnt/rice/default/Workspace/xuxiaolong/RNAprediction/RNA_bam/analysis/six_context_global_tmm_bigwig/bigwig/root_salt"
}

# 3. 初始化结果字典
sample_files = {sample: [] for sample in all_samples}

# 4. 遍历目录并匹配绝对路径
for condition, dir_path in dirs.items():
    if not os.path.exists(dir_path):
        print(f"警告: 路径不存在 -> {dir_path}")
        continue
        
    for root, _, files in os.walk(dir_path):
        for file in files:
            # 仅匹配 .bw 或 .bigwig 文件（可根据实际后缀调整）
            if file.endswith((".fasta", ".gff3", ".bw", ".bigwig")):
                full_path = os.path.abspath(os.path.join(root, file))
                
                # 遍历样本名称进行正则匹配
                for sample in all_samples:
                    # 只要文件名包含 sample，且 sample 后面跟的是非字母数字字符（如 _、.）或结尾，即可匹配
                    pattern = rf"{sample}(?=[^a-zA-Z0-9]|$)"
                    if re.search(pattern, file):
                        sample_files[sample].append(full_path)

# 打印最终构建的字典
import pprint
pprint.pprint(sample_files)
```

```python
def get_map(fasta_path): 
    import re
    import pandas as pd

    fasta_path = fasta_path

    gwh_to_chr = {}
    records = []

    with open(fasta_path, "r") as f:
        for line in f:
            if line.startswith(">"):
                # 提取 GWH ID (标头开头的 ID)
                gwh_id = line.split()[0].lstrip(">")

                # 使用正则精准提取 OriSeqID (如 Chr5)
                ori_match = re.search(r"OriSeqID=(\w+)", line)
                chr_name = ori_match.group(1) if ori_match else None

                # 备用匹配：如果没找到 OriSeqID，提取 Chromosome X
                if not chr_name:
                    chr_match = re.search(r"Chromosome\s+(\d+)", line)
                    if chr_match:
                        chr_name = f"Chr{chr_match.group(1)}"

                # 提取序列长度 Len
                len_match = re.search(r"Len=(\d+)", line)
                seq_len = int(len_match.group(1)) if len_match else None

                gwh_to_chr[gwh_id] = chr_name
                records.append(
                    {"GWH_ID": gwh_id, "Chromosome": chr_name, "Length": seq_len}
                )

    # 转换为 DataFrame
    df_mapping = pd.DataFrame(records)

    print(gwh_to_chr)

    # print(df_mapping)

    return gwh_to_chr
```

```python
import pyBigWig

def rename_bw_chromosomes_by_dict(input_bw_path, output_bw_path, gwh_to_chr):
    """
    根据 gwh_to_chr 字典将 BigWig 文件的染色体名称重命名（如 GWHDRAM00000008 -> Chr5），
    并另存为新的 BigWig 文件。
    """
    bw_in = pyBigWig.open(input_bw_path)
    old_chroms = bw_in.chroms()
    
    # 1. 根据 gwh_to_chr 映射生成新 Header
    old_to_new = {}
    new_header_dict = {}
    
    for old_chrom, length in old_chroms.items():
        if old_chrom in gwh_to_chr:
            new_chrom = gwh_to_chr[old_chrom]
            old_to_new[old_chrom] = new_chrom
            new_header_dict[new_chrom] = length
        else:
            print(f"警告: 染色体 {old_chrom} 在 gwh_to_chr 中未找到映射，将被跳过。")
            
    # pyBigWig 写入 Header 时建议按照染色体名称排序
    new_header = sorted(list(new_header_dict.items()), key=lambda x: x[0])
    
    if not new_header:
        bw_in.close()
        raise ValueError("没有可用的染色体映射关系，取消写出文件！")
    
    # 2. 创建并配置新 BigWig 文件
    bw_out = pyBigWig.open(output_bw_path, "w")
    bw_out.addHeader(new_header)
    
    # 3. 逐条染色体复制数据信号
    # 注意：需要按照新 Header 中排好序的染色体顺序依次写入
    for new_chrom, _ in new_header:
        # 找到对应的原 GWH ID
        old_chrom = [k for k, v in old_to_new.items() if v == new_chrom][0]
        length = old_chroms[old_chrom]
        
        intervals = bw_in.intervals(old_chrom, 0, length)
        if intervals:
            chroms_list = [new_chrom] * len(intervals)
            starts = [it[0] for it in intervals]
            ends = [it[1] for it in intervals]
            values = [it[2] for it in intervals]
            
            bw_out.addEntries(chroms_list, starts, ends=ends, values=values)
            
    bw_in.close()
    bw_out.close()
    print(f"转换成功: {input_bw_path} -> {output_bw_path}")
```
